# <font color = 'red'> Variables Dependientes

```python
credit_mix_mapping = {"Bad": 0, 
                       "Standard": 1, 
                       "Good": 2}

df["Credit_Score"] = df["Credit_Mix"].map(credit_mix_mapping)

df["Credit_Score"].unique()
```

```python
df["Binary_Credit_Score"] = df["Credit_Score"].apply(lambda x: 0 if x == 0 else 1)  # 0 = "Bad", 1 = "Not_Bad"
df["Binary_Credit_Score"].unique()
```

# <font color = 'red'> Dummy Variables

```python
# Crear variables dummy sin eliminar la variable categórica: algunas librerías procesan bien las variables categóricas
# otras prefieren las dummy. Por lo que se conservan las dos:

categorical_vars = [
    "Occupation", "Payment_of_Min_Amount", "Payment_Behaviour",
    "Last_Loan_1", "Last_Loan_2", "Last_Loan_3", "Last_Loan_4",
]

df_dummies_only = pd.get_dummies(df[categorical_vars], drop_first=False)

df_combined = pd.concat([df, df_dummies_only], axis=1)
```
