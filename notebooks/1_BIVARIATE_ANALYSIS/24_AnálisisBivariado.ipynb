## <font color = 'darkblue'> METODOLOGÍA

El análisis bivariado se lleva a cabo evaluando cada variable predictora por separado en relación con la variable a predecir. Se diferencian dos tipos de variables predictoras, aplicando técnicas específicas según su naturaleza.

Se diferencian dos tipos de variables en el análisis, aplicando metodologías específicas para cada caso.

* Variables Categóricas:

    - Se calcula la proporción de deudores en cada categoría del Credit_Score (Bad, Standard, Good) para cada valor de la variable categórica.

    - El objetivo es identificar categorías con mayor riesgo relativo y evaluar si dichas diferencias son razonables y consistentes con la lógica del negocio o fenómeno analizado.
    

* Variables Continuas:

    - Se dividen en deciles (10 grupos de igual tamaño según el valor de la variable).

    - Para cada decil, se calcula la distribución porcentual de Credit_Score (Bad, Standard, Good).

    - Este enfoque permite detectar relaciones crecientes o decrecientes (positivas o negativas) entre la variable continua y el nivel de riesgo crediticio.

Se documentan solo las relaciones bivariadas para las variables utilizadas en el modelo multivariado final.

```python
df = pd.read_csv(get_data_path("bivariate_preprocessed_data.csv"))
```

## <font color = 'darkblue'> PAYMENT OF MIN AMOUNT

```python
col = "Payment_of_Min_Amount"

category_names = [i for i in df.drop(col, axis = 1).columns if i.startswith(col)]

results = counts_and_proportions_by_category(df, category_names, "Credit_Mix")
counts_df = results["counts"]
proportions_df = results["proportions"]
summary_df = results["summary"]

counts_df.index = counts_df.index.str.replace(col + '_', '')
proportions_df.index = proportions_df.index.str.replace(col + '_', '')
summary_df.index = summary_df.index.str.replace(col + '_', '')
```

El comportamiento de pago del monto mínimo de la tarjeta de crédito genera una segmentación marcada entre los deudores:

* Dentro del grupo que no paga el monto mínimo, se observa que el 100% corresponde a deudores con buen historial crediticio: el 75% son clasificados como Good y el 25% como Standard. No hay deudores Bad en esta categoría.

* En contraste, entre quienes sí pagan el monto mínimo, no se registra ningún deudor Good. En este grupo, el 60% son Standard y el 40% son Bad.
     
De modo que no hay deudores buenos que paguen el monto mínimo. Ni malos que no lo paguen. Los Standard, en cambio, están en las dos clases
 pero principalmente en los que pagan el monto mínimo.

```python
custom_colors = {
    "prop_Bad": "orangered",
    "prop_Standard": "gold",
    "prop_Good": "skyblue"
}

fig = plot_categorical_proportions(
    df=proportions_df.reset_index(),  
    x_column="index",                 
    y_columns=["prop_Bad", "prop_Standard", "prop_Good"],  
    stacked=False,                      
    title=f"Proporción de Credit Score por {col}",
    x_title="",
    y_title="Proporción",
    legend_title="Credit Score",
    colors=custom_colors,              
    width=1000,
    height=600,
    sort_order='asc',          
    sort_by="prop_Bad"          
)

fig.show()
```

## <font color = 'darkblue'> Number of Credit Cards

Las medianas tienen el orden esperado: Median Bad > Mediana Standard > Mediana Good.

Esto concuerda con la hipótesis de que a mayor cantidad de tarjetas peor comportamiento de crédito.

```python
col = "Num_Credit_Card"

fig_box = px.box(df, x="Credit_Mix", y=col, title=f"Distribution of {col} by Credit Score Category", width=900, height=500,)
fig_box.show()
```

Los datos sugieren que a partir de siete tarjetas de crédito, la proporción de deudores Good disminuye de forma pronunciada, mientras que la proporción de deudores Bad muestra un incremento significativo. Este comportamiento indica que una alta tenencia de tarjetas podría estar asociada con un mayor riesgo crediticio.

```python
continuous_variable= col
decile_col_name = continuous_variable + '_Decile'
target_col_string = "Credit_Mix" # variable dependiente con nombres string
target_col = 'Credit_Score' # variable dependiente int (para modelos)

analysis_summary = summarize_decile_analysis(df, continuous_variable, decile_col_name, target_col_string)

# Obtener los resultados
df_deciles = analysis_summary["df_deciles"]  # DataFrame con los deciles asignados
deciles_summary = analysis_summary["decile_summary"]  # Resumen de deciles con conteos y proporciones

chart_types = {
    "prop_Good":"line",
    "prop_Standard":"line",
    "prop_Bad": "line",  
    "Decile_Count": "bar"    
}

fig = plot_interactive_chart(
    df=deciles_summary,  
    y_columns=["prop_Bad", "prop_Good", "prop_Standard", "Decile_Count"],  
    x_column="Decile_Max",  
    chart_types=chart_types, 
    title=f"Proportion of Credit Score Categories by {continuous_variable}",
    x_title="Decile",
    y_title="Decile Count",  
    y2_title="Proportion",   
    secondary_y=["prop_Bad", "prop_Good", "prop_Standard"],  
    width=900,
    height=500,
    custom_colors={"prop_Bad": "red", "prop_Good":"lightgreen",
    "prop_Standard":"brown", "Decile_Count": "gray"}  
)

fig.show()

display(round(deciles_summary.drop(['Decile_Proportion'], axis = 1), 3))
```
